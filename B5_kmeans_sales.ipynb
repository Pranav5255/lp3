```python
import pandas as pd, matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
df = pd.read_csv('sales_data_sample.csv', encoding='latin1')
X = StandardScaler().fit_transform(df[['QUANTITYORDERED','PRICEEACH','SALES']])
```

```python
wcss = [KMeans(k, n_init=10, random_state=0).fit(X).inertia_ for k in range(1,11)]
plt.plot(range(1,11), wcss, 'o-'); plt.xlabel('k'); plt.ylabel('WCSS'); plt.title('Elbow'); plt.show()
```

```python
km = KMeans(3, n_init=10, random_state=0).fit(X)   # k=3 from elbow
df['Cluster'] = km.labels_
plt.scatter(df.QUANTITYORDERED, df.SALES, c=df.Cluster); plt.xlabel('Qty'); plt.ylabel('Sales'); plt.show()
df.groupby('Cluster')[['QUANTITYORDERED','PRICEEACH','SALES']].mean()
```
