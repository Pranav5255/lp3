```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv('sales_data_sample.csv', encoding='latin1')
data = df[['QUANTITYORDERED', 'PRICEEACH', 'SALES']]

scaler = StandardScaler()
X = scaler.fit_transform(data)
```

```python
# Elbow method: plot WCSS for k = 1..10
wcss = []
for k in range(1, 11):
    km = KMeans(n_clusters=k, n_init=10, random_state=0)
    km.fit(X)
    wcss.append(km.inertia_)

plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel('Number of clusters (k)')
plt.ylabel('WCSS')
plt.title('Elbow Method')
plt.show()
```

```python
# Choose k where the curve bends (elbow) -> 3
kmeans = KMeans(n_clusters=3, n_init=10, random_state=0)
df['Cluster'] = kmeans.fit_predict(X)

plt.scatter(df['QUANTITYORDERED'], df['SALES'], c=df['Cluster'])
plt.xlabel('Quantity Ordered')
plt.ylabel('Sales')
plt.title('K-Means Clusters')
plt.show()

df.groupby('Cluster')[['QUANTITYORDERED', 'PRICEEACH', 'SALES']].mean()
```
