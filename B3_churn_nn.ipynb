```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
df = pd.read_csv('Churn_Modelling.csv')
df = df.drop(columns=['RowNumber', 'CustomerId', 'Surname'])          # useless columns
df = pd.get_dummies(df, columns=['Geography', 'Gender'], drop_first=True)  # text -> numbers

X = df.drop(columns=['Exited'])
y = df['Exited']                                                       # 1 = customer left
```

```python
# Split and normalize
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# Neural network: 2 hidden layers (16 and 8 neurons)
model = MLPClassifier(hidden_layer_sizes=(16, 8), max_iter=500, random_state=0)
model.fit(X_train, y_train)
pred = model.predict(X_test)

print('Accuracy:', accuracy_score(y_test, pred))
print('Confusion Matrix:')
print(confusion_matrix(y_test, pred))
```
