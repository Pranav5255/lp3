```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
df = pd.read_csv('Churn_Modelling.csv').drop(columns=['RowNumber','CustomerId','Surname'])
df = pd.get_dummies(df, columns=['Geography','Gender'], drop_first=True)
X, y = df.drop(columns='Exited'), df.Exited
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.2, random_state=0)
sc = StandardScaler().fit(Xtr); Xtr, Xte = sc.transform(Xtr), sc.transform(Xte)   # normalize
```

```python
m = MLPClassifier(hidden_layer_sizes=(16,8), max_iter=500, random_state=0).fit(Xtr, ytr)
p = m.predict(Xte)
print('Accuracy:', accuracy_score(yte,p)); print(confusion_matrix(yte,p))
```
