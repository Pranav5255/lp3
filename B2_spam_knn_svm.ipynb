```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
df = pd.read_csv('emails.csv')
X, y = df.drop(columns=['Email No.','Prediction']), df.Prediction
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.2, random_state=0)
sc = StandardScaler().fit(Xtr); Xtr, Xte = sc.transform(Xtr), sc.transform(Xte)
```

```python
for m in [KNeighborsClassifier(5), SVC(kernel='linear')]:
    p = m.fit(Xtr, ytr).predict(Xte)
    print(type(m).__name__, 'Accuracy:', accuracy_score(yte,p))
    print(confusion_matrix(yte,p)); print(classification_report(yte,p))
```
