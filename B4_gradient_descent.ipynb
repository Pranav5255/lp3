```python
import numpy as np, matplotlib.pyplot as plt
x, lr, xs = 2.0, 0.1, [2.0]
for i in range(100):
    x -= lr * 2*(x+3)          # dy/dx = 2(x+3)
    xs.append(x)
    if abs(xs[-1]-xs[-2]) < 1e-6: break
print('Local minimum at x = %.4f, y = %.6f, iterations = %d' % (x, (x+3)**2, i+1))
```

```python
t = np.linspace(-8, 3, 100)
plt.plot(t, (t+3)**2); plt.scatter(xs, [(v+3)**2 for v in xs], c='r', s=10); plt.show()
```
