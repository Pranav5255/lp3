```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return (x + 3) ** 2

def derivative(x):
    return 2 * (x + 3)
```

```python
x = 2                 # starting point
learning_rate = 0.1
precision = 0.000001
max_iters = 1000

history = [x]
for i in range(max_iters):
    new_x = x - learning_rate * derivative(x)
    history.append(new_x)
    if abs(new_x - x) < precision:
        break
    x = new_x

print('Local minimum at x =', round(x, 4))
print('y =', f(x))
print('Iterations:', i + 1)
```

```python
xs = np.linspace(-8, 3, 100)
plt.plot(xs, f(xs), label='y = (x+3)^2')
plt.scatter(history, [f(v) for v in history], color='red', s=10, label='steps')
plt.legend()
plt.show()
```
